# <span style="color: #a855f7">NumPy array operations</span>

<span style="color: #a855f7">створення масивів, avg, min, max, сортування, axis, зміна форми, булева індексація</span>

```python
from html import escape

import numpy as np
from IPython.display import HTML, Markdown, display


def colorize(text):
    display(HTML(f'<pre style="color: #a855f7">{escape(str(text))}</pre>'))


rng = np.random.default_rng(42)
np.set_printoptions(precision=4, linewidth=120)
```

## <span style="color: #a855f7">Random array and sorting</span>

<span style="color: #a855f7">10 випадкових цілих чисел від 0 до 100, min, max, avg і порівняння до та після сортування</span>

```python
numbers = rng.integers(0, 101, size=10)
min_number = np.min(numbers)
max_number = np.max(numbers)
avg_number = np.mean(numbers)
sorted_numbers = np.sort(numbers)

colorize(f"before: {numbers}\nafter:  {sorted_numbers}")
colorize(f"min: {min_number}\nmax: {max_number}\navg: {avg_number:.2f}")
```

## <span style="color: #a855f7">Row and column sums</span>

<span style="color: #a855f7">матриця 4×4 із цілих чисел від 1 до 50, суми рядків через axis=1, стовпців через axis=0 і загальне avg</span>

```python
matrix = rng.integers(1, 51, size=(4, 4))
row_sums = np.sum(matrix, axis=1)
column_sums = np.sum(matrix, axis=0)
avg_matrix = np.mean(matrix)

matrix_rows = "\n".join(
    "| " + " | ".join(
        f'<span style="color: #a855f7">{value}</span>' for value in row
    ) + " |"
    for row in matrix
)
display(Markdown(
    "| col 0 | col 1 | col 2 | col 3 |\n"
    "| ---: | ---: | ---: | ---: |\n" + matrix_rows
))
colorize(f"row sums: {row_sums}\ncolumn sums: {column_sums}\navg: {avg_matrix:.2f}")
```

## <span style="color: #a855f7">Transpose and reshape</span>

<span style="color: #a855f7">масив 3×5 із чисел від 0 до 14, транспонування міняє осі, reshape змінює форму з збереженням порядку елементів</span>

```python
original = np.arange(15).reshape(3, 5)
transposed = original.T
reshaped = original.reshape(5, 3)

colorize(f"original {original.shape}:\n{original}")
colorize(f"transposed {transposed.shape}:\n{transposed}")
colorize(f"reshaped {reshaped.shape}:\n{reshaped}")
```

## <span style="color: #a855f7">Min and max positions</span>

<span style="color: #a855f7">матриця 10×10 із випадкових чисел від 0 до 1, значення min і max, індекси у форматі row, column, індексація з 0</span>

```python
random_matrix = rng.random((10, 10))
min_value = np.min(random_matrix)
max_value = np.max(random_matrix)
min_position = tuple(int(index) for index in np.unravel_index(
    np.argmin(random_matrix), random_matrix.shape
))
max_position = tuple(int(index) for index in np.unravel_index(
    np.argmax(random_matrix), random_matrix.shape
))

colorize(random_matrix)
colorize(f"min: {min_value:.6f}, position: {min_position}")
colorize(f"max: {max_value:.6f}, position: {max_position}")
```

## <span style="color: #a855f7">Boolean indexing</span>

<span style="color: #a855f7">15 випадкових цілих чисел від -10 до 10, вибір додатних і парних значень за булевими масками</span>

```python
integers = rng.integers(-10, 11, size=15)
positive_numbers = integers[integers > 0]
even_numbers = integers[integers % 2 == 0]

colorize(f"original: {integers}")
colorize(f"positive: {positive_numbers}")
colorize(f"even: {even_numbers}")
```

## <span style="color: #a855f7">Concatenate and stack</span>

<span style="color: #a855f7">об’єднання двох масивів в один, в два рядки 2×5, в два стовпці 5×2</span>

```python
first = np.arange(1, 6)
second = np.arange(6, 11)
concatenated = np.concatenate((first, second))
stacked_rows = np.vstack((first, second))
stacked_columns = np.column_stack((first, second))

colorize(f"first: {first}\nsecond: {second}")
colorize(f"concatenated {concatenated.shape}:\n{concatenated}")
colorize(f"rows {stacked_rows.shape}:\n{stacked_rows}")
colorize(f"columns {stacked_columns.shape}:\n{stacked_columns}")
```
