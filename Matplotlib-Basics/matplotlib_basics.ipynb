# <span style="color: #a855f7">Matplotlib basics</span>

<span style="color: #a855f7">лінійні, стовпчасті, кругова, точкова діаграми, підписи, маркери, сітка</span>

```python
import matplotlib.pyplot as plt

PURPLE = "#a855f7"
YELLOW = "#facc15"
PINK = "#f472b6"
BLUE = "#60a5fa"

plt.style.use("dark_background")
plt.rcParams.update({
    "figure.figsize": (9, 4.5),
    "figure.dpi": 110,
    "figure.facecolor": "#111116",
    "axes.facecolor": "#111116",
    "axes.edgecolor": "#746a80",
    "axes.labelcolor": "#e9d5ff",
    "axes.titlecolor": "#e9d5ff",
    "axes.titlesize": 13,
    "axes.titlepad": 14,
    "text.color": "#e9d5ff",
    "xtick.color": "#d8c8ed",
    "ytick.color": "#d8c8ed",
    "grid.color": "#746a80",
    "grid.alpha": 0.25,
    "lines.linewidth": 2,
    "font.size": 10,
})
```

## <span style="color: #a855f7">Square function</span>

<span style="color: #a855f7">два списки: числа від 1 до 10, їхні квадрати, лінійний графік y = x² з маркером в кожній точці</span>

```python
numbers = list(range(1, 11))
squares = [number ** 2 for number in numbers]

fig, ax = plt.subplots(layout="constrained")
ax.plot(numbers, squares, color=PURPLE, marker="o")
ax.set(title="Function y = x²", xlabel="X", ylabel="Y")
ax.set_xticks(numbers)
ax.grid(axis="y")
ax.set_axisbelow(True)
plt.show()
```

## <span style="color: #a855f7">Monthly sales</span>

<span style="color: #a855f7">список 6 місяців і значення продажів, стовпчаста діаграма з нульовим початком осі Y</span>

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [150, 200, 250, 300, 280, 320]

fig, ax = plt.subplots(layout="constrained")
sales_bars = ax.bar(months, sales, color=BLUE)
ax.bar_label(sales_bars, padding=3, color=BLUE)
ax.set(title="Sales over six months", xlabel="Month", ylabel="Sales (units)")
ax.set_ylim(0, max(sales) * 1.15)
ax.grid(axis="y")
ax.set_axisbelow(True)
plt.show()
```

## <span style="color: #a855f7">Sales shares</span>

<span style="color: #a855f7">3 категорії з частками 40, 30 і 30</span>

```python
categories = ["Food", "Clothing", "Electronics"]
shares = [40, 30, 30]

fig, ax = plt.subplots(layout="constrained")
wedges, labels, percentages = ax.pie(
    shares,
    labels=categories,
    colors=[PURPLE, YELLOW, PINK],
    autopct="%1.0f%%",
    startangle=90,
    wedgeprops={"edgecolor": "#111116", "linewidth": 2},
)
for percentage in percentages:
    percentage.set_color("#111116")
    percentage.set_fontweight("bold")
ax.set_title("Sales share by category")
ax.set_aspect("equal")
plt.show()
```

## <span style="color: #a855f7">Height and weight</span>

<span style="color: #a855f7">5 пар значень, scatter зросту по X і ваги по Y</span>

```python
heights = [160, 165, 170, 175, 180]
weights = [55, 60, 70, 72, 80]

fig, ax = plt.subplots(layout="constrained")
ax.scatter(heights, weights, color=PINK, s=70)
ax.set(title="Weight vs height", xlabel="Height (cm)", ylabel="Weight (kg)")
ax.set_xticks(heights)
ax.grid()
ax.set_axisbelow(True)
plt.show()
```

## <span style="color: #a855f7">Temperature over time</span>

<span style="color: #a855f7">7 вимірювань за добу, лінійний графік температури за годинами</span>

```python
hours = [0, 4, 8, 12, 16, 20, 24]
temperature = [15, 14, 18, 22, 25, 20, 17]

fig, ax = plt.subplots(layout="constrained")
ax.plot(hours, temperature, color=PURPLE, marker="o")
ax.set(title="Temperature over 24 hours", xlabel="Hour", ylabel="Temperature (°C)")
ax.set_xticks(hours)
ax.grid()
ax.set_axisbelow(True)
plt.show()
```

## <span style="color: #a855f7">Test scores</span>

<span style="color: #a855f7">імена і результати 5 студентів, шкала від 0 до 100</span>

```python
students = ["Alex", "Irene", "John", "Mary", "Boris"]
scores = [85, 92, 78, 90, 88]

fig, ax = plt.subplots(layout="constrained")
score_bars = ax.bar(students, scores, color=YELLOW)
ax.bar_label(score_bars, padding=3, color=YELLOW)
ax.set(title="Test results", xlabel="Student", ylabel="Score (points)")
ax.set_ylim(0, 100)
ax.grid(axis="y")
ax.set_axisbelow(True)
plt.show()
```
