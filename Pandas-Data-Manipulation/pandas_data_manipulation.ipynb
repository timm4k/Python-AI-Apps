# <span style="color: #a855f7">Pandas data manipulation</span>

<span style="color: #a855f7">фільтрація, сортування, читання CSV, groupby, пошук max, merge

```python
from html import escape

import pandas as pd
from IPython.display import HTML, display


def colorize(value):
    if isinstance(value, pd.DataFrame):
        html = value.to_html(
            index=False, border=0, na_rep="missing", float_format="{:.2f}".format
        )
        html = html.replace("<table ", '<table align="left" ', 1)
        for tag in ("th", "td"):
            html = html.replace(f"<{tag}>", f'<{tag}><span style="color: #a855f7">')
            html = html.replace(f"</{tag}>", f"</span></{tag}>")
        display(HTML(html))
    else:
        display(HTML(f'<pre style="color: #a855f7">{escape(str(value))}</pre>'))
```

## <span style="color: #a855f7">Filtering and average price</span>

<span style="color: #a855f7">DataFrame з 6 рядками, avg ціни, фільтр year > 2015, сортування price за зростанням</span>

```python
books = pd.DataFrame([
    ("A Killer's Mind", "Mike Omer", 2018, 450),
    ("In the Darkness", "Mike Omer", 2019, 480),
    ("Thicker than Blood", "Mike Omer", 2020, 520),
    ("The Silent Patient", "Alex Michaelides", 2019, 430),
    ("The Girl on the Train", "Paula Hawkins", 2015, 390),
    ("The Devotion of Suspect X", "Keigo Higashino", 2011, 410),
], columns=["title", "author", "year", "price"])
avg_price = books["price"].mean()
recent_books = books.loc[books["year"] > 2015]
books_by_price = books.sort_values("price")

colorize(books)
colorize(f"avg price: {avg_price:.2f}")
colorize("year > 2015")
colorize(recent_books)
colorize("price ascending")
colorize(books_by_price)
```

## <span style="color: #a855f7">CSV and order statistics</span>

<span style="color: #a855f7">читання orders.csv, head(10), к-сть замовлень за customer через value_counts і min, max, sum для amount</span>

```python
csv_orders = pd.read_csv("orders.csv", parse_dates=["date"], date_format="%Y-%m-%d")
order_counts = csv_orders["customer"].value_counts().rename_axis("customer").reset_index(name="order_count")
amount_statistics = csv_orders["amount"].agg(["min", "max", "sum"]).rename_axis("metric").reset_index(name="amount")

colorize(csv_orders.head(10))
colorize(order_counts)
colorize(amount_statistics)
```

## <span style="color: #a855f7">Filtering and grouped averages</span>

<span style="color: #a855f7">DataFrame з 12 рядками, ккал і протеїн на 100г, фільтр ккал > 300, avg протеїн за категорією, сортування ккал за спаданням</span>

```python
foods = pd.DataFrame([
    ("Almonds", "Nuts", 579, 21.2),
    ("Walnuts", "Nuts", 654, 15.2),
    ("Peanuts", "Nuts", 567, 25.8),
    ("Oats", "Grains", 389, 16.9),
    ("Brown rice", "Grains", 370, 7.9),
    ("Buckwheat", "Grains", 343, 13.3),
    ("Greek yogurt", "Dairy", 59, 10.3),
    ("Cheddar", "Dairy", 403, 24.9),
    ("Lentils", "Legumes", 116, 9.0),
    ("Chickpeas", "Legumes", 164, 8.9),
    ("Banana", "Fruit", 89, 1.1),
    ("Avocado", "Fruit", 160, 2.0),
], columns=["product", "category", "calories", "protein"])
high_calorie_foods = foods.loc[foods["calories"] > 300]
category_protein = foods.groupby("category", as_index=False).agg(avg_protein=("protein", "mean"))
foods_by_calories = foods.sort_values("calories", ascending=False)

colorize(foods)
colorize("calories > 300")
colorize(high_calorie_foods)
colorize(category_protein)
colorize("calories descending")
colorize(foods_by_calories)
```

## <span style="color: #a855f7">Hours by employee and project</span>

<span style="color: #a855f7">початкова таблиця з 12 рядками, sum годин за name і за project, працівники з max загальною к-стю годин</span>

```python
work_logs = pd.DataFrame([
    ("Fiona Gallagher", "Website", 28),
    ("Lip Gallagher", "API", 32),
    ("Ian Gallagher", "Testing", 24),
    ("Debbie Gallagher", "Website", 18),
    ("Carl Gallagher", "Testing", 20),
    ("Veronica Fisher", "Analytics", 26),
    ("Kevin Ball", "Support", 22),
    ("Mickey Milkovich", "Support", 30),
    ("Fiona Gallagher", "Analytics", 16),
    ("Lip Gallagher", "Website", 20),
    ("Ian Gallagher", "API", 14),
    ("Debbie Gallagher", "Testing", 12),
], columns=["name", "project", "hours"])
employee_hours = work_logs.groupby("name", as_index=False).agg(total_hours=("hours", "sum"))
project_hours = work_logs.groupby("project", as_index=False).agg(total_hours=("hours", "sum"))
top_employees = employee_hours.loc[employee_hours["total_hours"] == employee_hours["total_hours"].max()]

colorize(work_logs)
colorize(employee_hours)
colorize(project_hours)
colorize("max total hours")
colorize(top_employees)
```

## <span style="color: #a855f7">Ticket sales aggregation</span>

<span style="color: #a855f7">DataFrame з 12 рядками, sum проданих квитків за movie і за city, фільми з max загальною к-стю квитків</span>

```python
ticket_sales = pd.DataFrame([
    ("The Wailing", "Seoul", 180),
    ("The Wailing", "Tokyo", 120),
    ("The Wailing", "Bangkok", 150),
    ("Cure", "Seoul", 110),
    ("Cure", "Tokyo", 160),
    ("Cure", "Bangkok", 90),
    ("Shutter", "Seoul", 140),
    ("Shutter", "Tokyo", 130),
    ("Shutter", "Bangkok", 210),
    ("A Tale of Two Sisters", "Seoul", 170),
    ("A Tale of Two Sisters", "Tokyo", 100),
    ("A Tale of Two Sisters", "Bangkok", 120),
], columns=["movie", "city", "tickets_sold"])
movie_totals = ticket_sales.groupby("movie", as_index=False).agg(total_tickets=("tickets_sold", "sum"))
city_totals = ticket_sales.groupby("city", as_index=False).agg(total_tickets=("tickets_sold", "sum"))
top_movies = movie_totals.loc[movie_totals["total_tickets"] == movie_totals["total_tickets"].max()]

colorize(ticket_sales)
colorize(movie_totals)
colorize(city_totals)
colorize("max ticket sales")
colorize(top_movies)
```

## <span style="color: #a855f7">Customer order totals</span>

<span style="color: #a855f7">6 клієнтів і 9 замовлень, left merge за id і customer_id з перевіркою one_to_many, sum вартості за клієнтом, клієнти без замовлень залишаються з total_cost = 0</span>

```python
customers = pd.DataFrame([
    (1, "Fiona Gallagher", "Chicago"),
    (2, "Lip Gallagher", "Chicago"),
    (3, "Ian Gallagher", "Chicago"),
    (4, "Debbie Gallagher", "Chicago"),
    (5, "Carl Gallagher", "Chicago"),
    (6, "Veronica Fisher", "Chicago"),
], columns=["id", "name", "city"])
customer_orders = pd.DataFrame([
    (3, "ORD-201", 650),
    (1, "ORD-202", 450),
    (5, "ORD-203", 390),
    (2, "ORD-204", 480),
    (1, "ORD-205", 520),
    (4, "ORD-206", 430),
    (3, "ORD-207", 410),
    (2, "ORD-208", 700),
    (1, "ORD-209", 300),
], columns=["customer_id", "order", "cost"])
customer_orders = customer_orders.astype({"customer_id": "Int64", "cost": "Int64"})

colorize(customers)
colorize(customer_orders)

merged_orders = customers.merge(
    customer_orders, left_on="id", right_on="customer_id", how="left", validate="one_to_many"
)
customer_totals = merged_orders.groupby(["id", "name", "city"], as_index=False).agg(total_cost=("cost", "sum"))

colorize(merged_orders)
colorize(customer_totals)
```
