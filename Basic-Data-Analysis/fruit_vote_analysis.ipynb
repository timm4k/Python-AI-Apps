# <span style="color: #a855f7">Аналіз голосування за фрукти</span>

<span style="color: #a855f7">підрахунок голосів, визначення лідера, окрема таблиця 3 найпопулярніших фруктів</span>

```python
from collections import Counter

votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]
```

```python
vote_counts = Counter(votes)
top_three = vote_counts.most_common(3)
leading_fruit, leading_votes = top_three[0]
```

```python
from IPython.display import Markdown, display


def colorize(text):
    lines = text.splitlines()
    for index, line in enumerate(lines):
        if index == 1:
            continue
        cells = line.strip("|").split("|")
        lines[index] = "| " + " | ".join(
            f'<span style="color: #a855f7">{cell.strip()}</span>' for cell in cells
        ) + " |"
    return "\n".join(lines)


vote_rows = "\n".join(
    f"| {fruit} | {count} |" for fruit, count in sorted(vote_counts.items())
)
display(Markdown(colorize("| Fruit | Votes |\n| --- | ---: |\n" + vote_rows)))
display(Markdown('<span style="color: #a855f7">' + f"<strong>Leader:</strong> {leading_fruit} — {leading_votes} votes" + '</span>'))
```

## <span style="color: #a855f7">Топ-3</span>

<span style="color: #a855f7">однакова к-сть голосів зберігає порядок першої появи фруктів у списку</span>

```python
top_rows = "\n".join(
    f"| {rank} | {fruit} | {count} |"
    for rank, (fruit, count) in enumerate(top_three, start=1)
)
display(Markdown(colorize("| Rank | Fruit | Votes |\n| ---: | --- | ---: |\n" + top_rows)))
```
