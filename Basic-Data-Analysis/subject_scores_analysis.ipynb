# <span style="color: #a855f7">Порівняння результатів із предметів</span>

<span style="color: #a855f7">avg, min і max бали для кожного предмета, автоматичне визначення найвищого avg результату</span>

```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]
```

```python
subject_scores = {"Math": math, "Physics": physics, "English": english}
subject_statistics = {
    subject: {
        "average": sum(scores) / len(scores),
        "minimum": min(scores),
        "maximum": max(scores),
    }
    for subject, scores in subject_scores.items()
}
best_subject = max(
    subject_statistics,
    key=lambda subject: subject_statistics[subject]["average"],
)
```

```python
from IPython.display import Markdown, display


def colorize(text):
    lines = text.splitlines()
    for index, line in enumerate(lines):
        if index == 1:
            continue
        cells = line.strip("|").split("|")
        lines[index] = "| " + " | ".join(
            f'<span style="color: #a855f7">{cell.strip()}</span>' for cell in cells
        ) + " |"
    return "\n".join(lines)


score_rows = "\n".join(
    f"| {subject} | {statistics['average']:.2f} | {statistics['minimum']} | {statistics['maximum']} |"
    for subject, statistics in subject_statistics.items()
)
display(Markdown(
    colorize("| Subject | Average | Minimum | Maximum |\n"
    "| --- | ---: | ---: | ---: |\n" + score_rows)
))
```

```python
display(Markdown('<span style="color: #a855f7">' + f"<strong>Subject with the highest average score:</strong> {best_subject}" + '</span>'))
```
