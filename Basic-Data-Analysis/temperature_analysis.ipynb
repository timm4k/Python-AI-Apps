# <span style="color: #a855f7">Аналіз температур</span>

<span style="color: #a855f7">обчислення avg температури, min, max і range за сім спостережень</span>

```python
temperatures = [2, 5, 4, 6, 7, 3, 1]
```

```python
average_temperature = sum(temperatures) / len(temperatures)
minimum_temperature = min(temperatures)
maximum_temperature = max(temperatures)
temperature_range = maximum_temperature - minimum_temperature
```

```python
from IPython.display import Markdown, display


def colorize(text):
    lines = text.splitlines()
    for index, line in enumerate(lines):
        if index == 1:
            continue
        cells = line.strip("|").split("|")
        lines[index] = "| " + " | ".join(
            f'<span style="color: #a855f7">{cell.strip()}</span>' for cell in cells
        ) + " |"
    return "\n".join(lines)


display(Markdown(colorize(
    "| Metric | Value |\n"
    "| --- | ---: |\n"
    f"| Average temperature | {average_temperature:.2f} |\n"
    f"| Minimum temperature | {minimum_temperature} |\n"
    f"| Maximum temperature | {maximum_temperature} |\n"
    f"| Temperature range | {temperature_range} |"
)))
```
