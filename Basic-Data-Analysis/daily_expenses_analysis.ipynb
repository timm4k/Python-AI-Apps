# <span style="color: #a855f7">Аналіз щоденних витрат</span>

<span style="color: #a855f7">total, avg, min і max витрата, кількість днів з витратами більше 1000</span>

```python
daily_expenses = [1200, 1500, 800, 600, 1000, 400, 900]
```

```python
total_expenses = sum(daily_expenses)
average_expense = total_expenses / len(daily_expenses)
minimum_expense = min(daily_expenses)
maximum_expense = max(daily_expenses)
days_above_1000 = sum(expense > 1000 for expense in daily_expenses)
```

```python
from IPython.display import Markdown, display


def colorize(text):
    lines = text.splitlines()
    for index, line in enumerate(lines):
        if index == 1:
            continue
        cells = line.strip("|").split("|")
        lines[index] = "| " + " | ".join(
            f'<span style="color: #a855f7">{cell.strip()}</span>' for cell in cells
        ) + " |"
    return "\n".join(lines)


display(Markdown(colorize(
    "| Metric | Value |\n"
    "| --- | ---: |\n"
    f"| Total expenses | {total_expenses} |\n"
    f"| Average daily expense | {average_expense:.2f} |\n"
    f"| Minimum daily expense | {minimum_expense} |\n"
    f"| Maximum daily expense | {maximum_expense} |\n"
    f"| Days with expenses above 1000 | {days_above_1000} |"
)))
```
