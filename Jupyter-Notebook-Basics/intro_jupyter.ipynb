## <span style="color: #a855f7">Arithmetic</span>

<span style="color: #a855f7">результат розрахунків sum і diff</span>

```python
a = 15
b = 7

sum_result = a + b
difference = a - b

colorize(f"sum: {sum_result}")
colorize(f"difference: {difference}")
```

<span style="color: #a855f7">такий ж приклад для product і quotient</span>

```python
a = 15
b = 7

product = a * b
quotient = a / b

colorize(f"product: {product}")
colorize(f"quotient: {quotient:.2f}")
```

## <span style="color: #a855f7">Student scores</span>

<span style="color: #a855f7">avg, max і min бали групи, студенти з балом нижче 60</span>

<span style="color: #a855f7">для кожного студента видно зірки за кількістю десятків у балі</span>

```python
students = [
    ("Obsidia", 85),
    ("Selena", 92),
    ("Alice", 74),
    ("Dyrroth", 60),
    ("Helcurt", 48),
]
```

```python
scores = [score for name, score in students]
avg_score = sum(scores) / len(scores)
max_score = max(scores)
min_score = min(scores)
students_below_60 = [(name, score) for name, score in students if score < 60]

colorize("below 60: " + ", ".join(
    f"{name} ({score})" for name, score in students_below_60
))
```

### <span style="color: #a855f7">Score chart</span>

<span style="color: #a855f7">одна зірка відповідає десяти балам, залишок відкидається</span>

```python
for name, score in students:
    stars = "*" * (score // 10)
    colorize(f"{name} | {stars}")
```
